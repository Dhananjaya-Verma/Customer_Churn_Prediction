```python
# ========================================
# Telco Customer Churn - SVM Prediction Script
# ========================================

import pandas as pd
import joblib
import os

# -------------------- Output Path --------------------
output_path = r"D:\CSE(DataScience)\Customer_Churn_Prediction\outputs"
os.makedirs(output_path, exist_ok=True)

# -------------------- Load Trained Model --------------------
# Pehle tu apna trained model save karna hoga training ke baad
# Example (from training script):
# joblib.dump(svm_pipeline, "models/svm_pipeline.pkl")

model_path = "D:\CSE(DataScience)\Customer_Churn_Prediction\models\churn_pipeline_svm.pkl"
svm_pipeline = joblib.load(model_path)

# -------------------- Example Input --------------------
# Single customer data (dictionary format)
new_customer = {
    "gender": "Female",
    "SeniorCitizen": 0,
    "Partner": "Yes",
    "Dependents": "No",
    "tenure": 12,
    "PhoneService": "Yes",
    "MultipleLines": "No",
    "InternetService": "Fiber optic",
    "OnlineSecurity": "No",
    "OnlineBackup": "Yes",
    "DeviceProtection": "No",
    "TechSupport": "No",
    "StreamingTV": "Yes",
    "StreamingMovies": "Yes",
    "Contract": "Month-to-month",
    "PaperlessBilling": "Yes",
    "PaymentMethod": "Electronic check",
    "MonthlyCharges": 70.35,
    "TotalCharges": 845.5
}

# Convert to DataFrame
new_customer_df = pd.DataFrame([new_customer])

# -------------------- Make Prediction --------------------
prediction = svm_pipeline.predict(new_customer_df)[0]
probability = svm_pipeline.predict_proba(new_customer_df)[0][1]

result = {
    "Prediction": "Churn" if prediction == 1 else "Not Churn",
    "Churn_Probability": probability
}

# -------------------- Save Prediction --------------------
result_df = pd.DataFrame([result])
result_path = os.path.join(output_path, "svm_single_prediction.csv")
result_df.to_csv(result_path, index=False)

print("✅ Prediction saved at:", result_path)
print(result)
```
